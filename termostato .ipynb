```python
Diseño de agente (Ficha PEAS)
```

Percepción (S) : Lectura de la temperatura actual del ambiente (medida en grados Celsius por el sensor).
Acciones (A) : Activar calefacción, Activar aire acondicionado, o Apagar climatizador / Mantener apagado.
Entorno (E) : Habitación u oficina cerrada donde varía la temperatura ambiente según el clima externo.
Objetivo: Mantener la temperatura de la habitación dentro de un rango confortable (cerca de la temperatura deseada).
Medida de desempeño (P): Tiempo total que la habitación se mantiene en la temperatura ideal y consumo mínimo de energía.

```python
de PEAS a reglas de decision
```

```python
Si la temperatura está muy por encima de la deseada (diferencia > margen): Enfriar (Activar aire acondicionado)
Si la temperatura está muy por debajo de la deseada (diferencia < -margen): Calentar (Activar calefacción)
Si la temperatura está cerca de la deseada (-margen <= diferencia <= margen): Apagar (Mantener sistema apagado / reposo)
```

Implementación

```python
def agente_termostato(temperatura_actual, temperatura_objetivo=22.0, margen=1.0):
    """
    Agente reactivo simple para control de climatización.
    Percibe la temperatura actual y decide accionar calefacción,
    aire acondicionado o apagarse según el margen configurado.
    """
    diferencia = temperatura_actual - temperatura_objetivo
    
    if diferencia > margen:
        return "Activar Aire Acondicionado"
    elif diferencia < -margen:
        return "Activar Calefacción"
    else:
        return "Apagar Climatizador"
```

Prueba

```python
# Percepciones base + 2 adicionales (22.0 como valor exacto y 23.0 como límite del margen)
percepciones = [18.0, 19.5, 21.8, 24.3, 26.0, 23.0, 20.5, 22.0, 23.0]

for t in percepciones:
    accion = agente_termostato(t)
    print(f"Percepción: {t}°C -> Acción del agente: {accion}")
```

Reflexión

¿Es racional este agente termostato?
Sí, es racional dentro del alcance de su diseño reactivo, ya que siempre toma la mejor decisión inmediata para acercar la temperatura al rango objetivo basándose en la percepción recibida.
